```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("UP_rainfall_dataset.zip")
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
district_rain = df.groupby("DISTRICT")["PRECTOTCORR"].mean().sort_values(ascending=False)
district_rain
```

```python
district_rain.head(10)
```

```python
monthly_rain = df.groupby("MO")["PRECTOTCORR"].mean()
monthly_rain
```

```python
yearly_rain = df.groupby("YEAR")["PRECTOTCORR"].mean()
yearly_rain
```

```python
print("Maximum Rainfall:", df["PRECTOTCORR"].max())
print("Minimum Rainfall:", df["PRECTOTCORR"].min())
print("Average Rainfall:", df["PRECTOTCORR"].mean())
print("Median Rainfall:", df["PRECTOTCORR"].median())
print("Standard Deviation:", df["PRECTOTCORR"].std())
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df["PRECTOTCORR"])
plt.title("Outliers in Rainfall")
plt.show()
```

```python
Q1 = df["PRECTOTCORR"].quantile(0.25)
Q3 = df["PRECTOTCORR"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["PRECTOTCORR"] < lower) | (df["PRECTOTCORR"] > upper)]

print("Number of Outliers:", outliers.shape[0])
print("EDA completed")
```
