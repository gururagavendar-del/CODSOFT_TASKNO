```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv("UP_rainfall_dataset.zip")
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["PRECTOTCORR"], bins=30, edgecolor="black")
plt.title("Distribution of Rainfall")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Frequency")
plt.show()
```

```python
top10 = df.groupby("DISTRICT")["PRECTOTCORR"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top10.plot(kind="bar")
plt.title("Top 10 Districts by Average Rainfall")
plt.xlabel("District")
plt.ylabel("Average Rainfall (mm)")
plt.xticks(rotation=45)
plt.show()
```

```python
yearly = df.groupby("YEAR")["PRECTOTCORR"].mean()

plt.figure(figsize=(10,5))
plt.plot(yearly.index, yearly.values, marker='o')
plt.title("Year-wise Average Rainfall")
plt.xlabel("Year")
plt.ylabel("Average Rainfall (mm)")
plt.grid(True)
plt.show()
```

```python
top5 = df.groupby("DISTRICT")["PRECTOTCORR"].sum().sort_values(ascending=False).head(5)

plt.figure(figsize=(7,7))
plt.pie(top5.values,
        labels=top5.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Top 5 Districts by Total Rainfall")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["T2M_MAX"], df["PRECTOTCORR"], alpha=0.5)
plt.title("Maximum Temperature vs Rainfall")
plt.xlabel("Maximum Temperature")
plt.ylabel("Rainfall")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["PRECTOTCORR"])
plt.title("Box Plot of Rainfall")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(
    data=df,
    y="DISTRICT",
    order=df["DISTRICT"].value_counts().head(10).index
)

plt.title("Top 10 Districts by Number of Records")
plt.xlabel("Count")
plt.ylabel("District")
plt.show()
```

```python
monthly = df.groupby("MO")["PRECTOTCORR"].mean()

plt.figure(figsize=(10,5))
monthly.plot(kind="bar")
plt.title("Monthly Average Rainfall")
plt.xlabel("Month")
plt.ylabel("Average Rainfall (mm)")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10,5))
yearly.plot()
plt.title("Year-wise Average Rainfall")
plt.xlabel("Year")
plt.ylabel("Average Rainfall")
plt.grid(True)

plt.savefig("Yearly_Rainfall.png")
plt.show()
```
